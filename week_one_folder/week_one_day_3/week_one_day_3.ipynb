```
#Write a function to find the largest element in a list without using max(). 2

numbers = [45, 12, 78, 90, 12, 56, 23, 45]
def find_largest(numbers):
    largest = numbers[0]
    for i in numbers:
        if i > largest:
            largest = i
    return largest
print(find_largest(numbers))
```

```
#Write a function that counts how many even numbers are present in a list. 3
numbers = [45, 12, 78, 90, 12, 56, 23, 45]
def count_even(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0:
            count += 1
    return count
print("Even numbers count:", count_even(numbers))
```

```
#6. Palindrome List
#Write a function to check whether a list reads the same forward and backward.
numbers =  [45, 12, 78, 90, 12, 56, 23, 45]
def palindrome(numbers):
  rev_list = []
  for i in range(len(numbers)-1, -1, -1):
    rev_list.append(numbers[i])
  if rev_list == numbers:
    return "Palindrome num"
  else:
    return "not a palindrome num"
print(palindrome(numbers))
```

```
 # 8 Duplicate Elements
#Write a function that identifies duplicate elements in a list.
numbers = [45, 12, 78, 90, 12, 56, 23, 45]
def find_duplicates(numbers):
  duplicates = []
  for i in numbers:
    if numbers.count(i) > 1 and i not in duplicates:
      duplicates.append(i)
  return duplicates
print(find_duplicates(numbers))
```

```
# 9 . Function for Student Marks
#Create a function that accepts student marks in a list and returns average, highest, and lowest marks.

numbers = [45, 12, 78, 90, 12, 56, 23, 45]
def student_marks(numbers):
    total = sum(numbers)
    avg = total / len(numbers)
    highest = max(numbers)
    lowest = min(numbers)
    return avg, highest, lowest
result = student_marks(numbers)
print("Average:", result[0])
print("Highest:", result[1])
print("Lowest:", result[2])
```

```
#7. Second Largest Number
#Write a function to find the second largest number in a list.
numbers = [45, 12, 78, 90, 12, 56, 23, 45]
def second_largest(numbers):
    largest = numbers[0]
    second = numbers[1]
    for i in numbers:
        if i > largest:
            second = largest
            largest = i
        elif i > second and i != largest:
            second = i
    return second
print(second_largest(numbers))
```
