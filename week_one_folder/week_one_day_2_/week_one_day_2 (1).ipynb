```python
#Q12. Simulate payment retry logic where the system retries payment a maximum of 3 times if payment fails. easy one simple
attempt = 0

while attempt < 3:
    status = input("status: ")

    if status == "success":
        print("Payment Successful")
        break
    else:
        print("Payment Failed")
        attempt += 1

    if attempt == 3:
        print("Max retries reached")
```

```python
# 13 Write a program that keeps checking order status until status becomes 'Delivered' or maximum 5 attempts are reached.
attempt = 0
status = "processing"

while attempt < 5:
    print("Checking status...")

    if status == "Delivered":
        print("Delivered")
        break
    else:
        print("Not Delivered")

    attempt += 1

if attempt == 5:
    print("Max attempts reached")
```

```python
orders = [
    {"order_id": 1, "amount": 120000, "status": "Completed"},
    {"order_id": 2, "amount": 45000, "status": "Cancelled"},
    {"order_id": 3, "amount": 80000, "status": "Completed"},
    {"order_id": 4, "amount": 150000, "status": "Completed"}
]

premium_count = 0

for order in orders:

    if order["status"] == "Cancelled":
        continue

    if order["amount"] > 100000:
        print(order["order_id"], "- Premium Order")
        premium_count += 1
    else:
        print(order["order_id"], "- Regular Order")

print("Total Premium Orders:", premium_count)
print("Total Premium Orders:", premium_count)
```
