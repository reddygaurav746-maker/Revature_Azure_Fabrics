Medical Distribution System - Plotting Assignment

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

medical_data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Medicine_Distributed': [1200, 1500, 1800, 1700, 2100, 2500],
    'Hospital_Orders': [30, 45, 50, 48, 60, 75],
    'Delivery_Time': [5, 4, 6, 5, 3, 2],
    'Revenue': [50000, 65000, 72000, 70000, 85000, 95000]
}

df = pd.DataFrame(medical_data)

print(df)
```

```python
# 1Create a line chart showing medicine distribution over months.
plt.plot(df['Month'], df['Medicine_Distributed'], marker='o')

plt.title("Medicine Distribution Over Months")
plt.xlabel("Month")
plt.ylabel("Medicine Distributed")
plt.grid()

plt.savefig("medicine_distribution.png")
plt.show()
```

```python
# 3 Create a line chart for monthly revenue growth.

plt.plot(df['Month'], df['Revenue'], marker='o')

plt.title("Monthly Revenue Growth")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid()

plt.savefig("revenue_growth.png")
plt.show()
```

```python
   #4 Create multiple line charts for Medicine_Distributed and Hospital_Orders.
plt.plot(df['Month'], df['Medicine_Distributed'], marker='o', label='Medicine Distributed')

plt.plot(df['Month'], df['Hospital_Orders'], marker='o', label='Hospital Orders')

plt.title("Medicine Distributed vs Hospital Orders")
plt.xlabel("Month")
plt.ylabel("Values")

plt.legend()
plt.grid()

plt.savefig("multiple_line_chart.png")
plt.show()
```

```python
# Q5: Create a vertical bar chart for hospital orders by month

import matplotlib.pyplot as plt

plt.figure()
plt.bar(df['Month'], df['Hospital_Orders'])

plt.title("Hospital Orders by Month")
plt.xlabel("Month")
plt.ylabel("Hospital Orders")

plt.grid(True)
plt.show()
```

```python
# Q6: Create a horizontal bar chart for revenue by month

import matplotlib.pyplot as plt

plt.figure()
plt.barh(df['Month'], df['Revenue'])

plt.title("Revenue by Month")
plt.xlabel("Revenue")
plt.ylabel("Month")

plt.grid(True)
plt.show()# Q: Create a horizontal bar chart for revenue by month

import matplotlib.pyplot as plt

plt.figure()
plt.barh(df['Month'], df['Revenue'])

plt.title("Revenue by Month")
plt.xlabel("Revenue")
plt.ylabel("Month")

plt.grid(True)
plt.show()
```

```python
# Q7: Create a grouped bar chart comparing Medicine_Distributed and Hospital_Orders

import numpy as np
import matplotlib.pyplot as plt

x = np.arange(len(df['Month']))
width = 0.4

plt.figure()

plt.bar(x, df['Medicine_Distributed'], width=width, label="Medicine Distributed")
plt.bar(x + width, df['Hospital_Orders'], width=width, label="Hospital Orders")

plt.title("Medicine Distributed vs Hospital Orders")
plt.xlabel("Month")
plt.ylabel("Values")

plt.xticks(x + width/2, df['Month'])
plt.legend()
plt.grid(True)

plt.show()
```
