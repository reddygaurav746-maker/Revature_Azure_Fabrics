```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("RDDAssignment") \
    .getOrCreate()

sc = spark.sparkContext

print("Spark Started Successfully")
```

```python
  emp_rdd = sc.parallelize([
      (101,"John","IT",50000),
      (102,"Alice","HR",60000),
      (103,"Bob","IT",70000),
      (104,"David","Finance",80000),
      (105,"Eva","HR",55000),
      (106,"Frank","IT",65000),
      (107,"Grace","Finance",90000),
      (108,"Helen","HR",62000)
  ])

  print(emp_rdd.collect())
```

```python
# Question 1: Display all employee records using collect()

all_employees = emp_rdd.collect()
print(all_employees)
```

```python
# Question 2: Store collect() result in a variable and print it

employee_data = emp_rdd.collect()
print(employee_data)
```

```python
# Question 3: Display only employee names

names = emp_rdd.map(lambda x: x[1]).collect()
print(names)
```

```python
# Question 4: Collect all Finance department employees

finance_employees = emp_rdd.filter(lambda x: x[2] == "Finance").collect()
print(finance_employees)
```

```python
# Question 5: Collect employees earning more than 70000

high_salary = emp_rdd.filter(lambda x: x[3] > 70000).collect()
print(high_salary)
```

```python
# Question 6: Collect all unique department names

departments = emp_rdd.map(lambda x: x[2]).distinct().collect()
print(departments)
```

```python
# Question 7: Collect employee IDs and salaries

id_salary = emp_rdd.map(lambda x: (x[0], x[3])).collect()
print(id_salary)
```

```python
# Question 8: Collect employees sorted by salary

sorted_salary = emp_rdd.sortBy(lambda x: x[3]).collect()
print(sorted_salary)
```

```python
# Question 9: Collect only HR employee names

hr_names = emp_rdd.filter(lambda x: x[2] == "HR") \
                  .map(lambda x: x[1]) \
                  .collect()

print(hr_names)
```

```python
# Question 10: Increase salary by 10%

updated_salary = emp_rdd.map(
    lambda x: (x[0], x[1], x[2], x[3] * 1.10)
).collect()

print(updated_salary)
```
