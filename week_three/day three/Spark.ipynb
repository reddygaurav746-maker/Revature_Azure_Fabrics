```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

spark = SparkSession.builder.appName("EmployeeDFAssignment").getOrCreate()

data = [
    (101, "John", "IT", 50000, "Chennai", 30),
    (102, "Alice", "HR", 60000, "Bangalore", 28),
    (103, "Bob", "Finance", 70000, "Mumbai", 35),
    (104, "David", "IT", 80000, "Chennai", 40),
    (105, "Eva", "HR", 55000, "Hyderabad", 25),
    (106, "Frank", "Finance", 90000, "Mumbai", 45)
]

columns = ["emp_id", "emp_name", "department", "salary", "city", "age"]

emp_df = spark.createDataFrame(data, columns)
```

Section 7 – Null Handling
Create DataFrame with NULL salary

```python
data_null = [
    (101,"John","IT",50000,"Chennai",30),
    (102,"Alice","HR",None,"Bangalore",28),
    (103,"Bob","Finance",70000,"Mumbai",35),
    (104,"David","IT",80000,"Chennai",40)
]

emp_null_df = spark.createDataFrame(
    data_null,
    columns
)

emp_null_df.show()
```

1. Records having NULL salaries

```python
emp_null_df.filter(col("salary").isNull()).show()
```

2. Replace NULL salary with ₹50,000

```python
emp_null_df.fillna({"salary":50000}).show()
```

3. Drop records containing NULL values

```python
emp_null_df.na.drop().show()
```

4. Count NULL salary records

```python
emp_null_df.filter(
    col("salary").isNull()
).count()
```

5. Replace NULL salary using department average salary

```python
dept_avg = emp_null_df.groupBy("department") \
                      .agg(avg("salary").alias("avg_salary"))

result = emp_null_df.join(
    dept_avg,
    "department",
    "left"
).withColumn(
    "salary",
    when(col("salary").isNull(), col("avg_salary"))
    .otherwise(col("salary"))
).drop("avg_salary")

result.show()
```
